```python
import matplotlib.pyplot as plt

import numpy as np
```

```python
# Fix a seed for reproducibility
np.random.seed(19680801)

x = np.random.randn(1000)
y = np.random.randn(1000)

# Calculate bins for histogram plotting:
binwidth = 0.25
xymax = max(np.max(np.abs(x)), np.max(np.abs(y)))
lim = (int(xymax/binwidth) + 1) * binwidth
bins = np.arange(-lim, lim + binwidth, binwidth)
```

```python
fig, ax = plt.subplots()
scatter = ax.scatter(x, y)
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2)

scatter = ax1.scatter(x, y)
histy = ax2.hist(y, orientation='horizontal')
```

```python
fig, axs = plt.subplot_mosaic([['histx', '.'],
                               ['scatter', 'histy']],
                               layout='constrained')

scatter = axs['scatter'].scatter(x, y)
histx = axs['histx'].hist(x)
histy = axs['histy'].hist(y, orientation='horizontal')
```

```python
fig, axs = plt.subplot_mosaic([['histx', '.'],
                               ['scatter', 'histy']],
                               figsize=(6, 6),
                               width_ratios=(4, 1),
                               height_ratios=(1, 4),
                               layout='constrained')

scatter = axs['scatter'].scatter(x, y)
histx = axs['histx'].hist(x, bins=bins)
histy = axs['histy'].hist(y, bins=bins, orientation='horizontal')
```

```python
fig, axs = plt.subplot_mosaic([['histx', '.'],
                               ['scatter', 'histy']],
                               figsize=(6, 6),
                               width_ratios=(4, 1),
                               height_ratios=(1, 4),
                               layout='constrained')

scatter = axs['scatter'].scatter(x, y)
histx = axs['histx'].hist(x, bins=bins)
histy = axs['histy'].hist(y, bins=bins, orientation='horizontal')

axs['histx'].set_xticks([])
axs['histy'].set_yticks([])
```
