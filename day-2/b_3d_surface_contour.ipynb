```python
import matplotlib.pyplot as plt
from matplotlib import cm
from matplotlib.ticker import LinearLocator

import numpy as np
```

```python
x_ticks = np.arange(-5, 5, 0.25)
y_ticks = np.arange(-5, 5, 0.25)

X, Y = np.meshgrid(x_ticks, y_ticks)

R = np.sqrt(X**2 + Y**2)
Z = np.sin(R)
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, subplot_kw={"projection": "3d"})

surf_r = ax1.plot_surface(X, Y, R)
surf_z = ax2.plot_surface(X, Y, Z)

ax1.set_title('Z = R')
ax2.set_title('Z = sin(R)')
```

```python
# There's a jump in complication here from the last code cell;
# may want to split these steps up or edit/remove some parts of them

fig, ax = plt.subplots(subplot_kw={"projection": "3d"})

surf = ax.plot_surface(X, Y, Z,
                       cmap=cm.coolwarm,
                       linewidth=0,
                       antialiased=False)

ax.set_zlim(-1.01, 1.01)
ax.zaxis.set_major_locator(LinearLocator(10))
ax.zaxis.set_major_formatter('{x:.02f}')

fig.colorbar(surf, shrink=0.5, aspect=5)
```

[Colormaps](https://matplotlib.org/stable/users/explain/colors/colormaps.html)

```python
# This is the same exact plot as in the last code cell but with a differnt colormap
# Maybe move this to supplemental or show colormaps somewhere/how else

fig, ax = plt.subplots(subplot_kw={"projection": "3d"})

surf = ax.plot_surface(X, Y, Z,
                       cmap=cm.magma,
                       linewidth=0,
                       antialiased=False)

ax.set_zlim(-1.01, 1.01)
ax.zaxis.set_major_locator(LinearLocator(10))
ax.zaxis.set_major_formatter('{x:.02f}')

fig.colorbar(surf, shrink=0.5, aspect=5)
```
