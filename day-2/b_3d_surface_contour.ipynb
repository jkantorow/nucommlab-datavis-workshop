First we include a cell to import the packages we'll be using:

```python
import matplotlib.pyplot as plt

import numpy as np
```

Here we define the data we'd like to plot. Numpy's `np.arange()` function gives us an evenly distributed array of values between a maximum and minimum, while `np.meshgrid()` generates a grid of points from 2 or more input arrays.

```python
x_ticks = np.arange(-5, 5, 0.25)
y_ticks = np.arange(-5, 5, 0.25)

X, Y = np.meshgrid(x_ticks, y_ticks)

R = np.sqrt(X**2 + Y**2)
Z = np.sin(R)
```

Calling `plt.subplots()` allows us to separate the figure space and plotted axes into two separate objects (`fig` and `ax`, respectively). This is useful for when multiple plots (`ax` objects) need to be displayed on a single figure (`fig` object), or when plotting in 3 dimensions. In this example, we show one plot on the figure as a 3D surface contour using `ax.plot_surface()`.

```python
fig, ax = plt.subplots(subplot_kw={"projection": "3d"})

surf = ax.plot_surface(X, Y, Z)
```

We can give the surface a custom [colormap](https://matplotlib.org/stable/users/explain/colors/colormaps.html) by adding the `cmap` keyword argument to `ax.plot_surface()`. This can help visually denote the surface's z values and better show the 3 dimensions of the plot.

```python
fig, ax = plt.subplots(subplot_kw={"projection": "3d"})

surf = ax.plot_surface(X, Y, Z,
                       cmap='coolwarm')
```

Setting the `antialiased` keyword argument of `ax.plot_surface()` to `False` shows the surface as opaque which removes the wireframe lines from the image.

```python
fig, ax = plt.subplots(subplot_kw={"projection": "3d"})

surf = ax.plot_surface(X, Y, Z,
                       cmap='coolwarm',
                       antialiased=False)
```

We can give the axes a more streamlined look by removing the z-ticks and adding a colorbar to represent the surface z values instead.

```python
fig, ax = plt.subplots(subplot_kw={"projection": "3d"})

surf = ax.plot_surface(X, Y, Z,
                       cmap='coolwarm',
                       antialiased=False)

ax.set_zticklabels([]) # remove the z-axis tick labels but keep the grid lines

# Add a colorbar detailing the numerical range of the contour colors:
cbar = ax.figure.colorbar(surf, ax=ax, shrink=0.5, aspect=5)
cbar.ax.set_ylabel('Amplitude', rotation=-90, va='bottom') # add a label for the colorbar
```

We can explicitly display the maximum and minimum surface z values on the colorbar without making it too crowded.

```python
fig, ax = plt.subplots(subplot_kw={"projection": "3d"})

surf = ax.plot_surface(X, Y, Z,
                       cmap='coolwarm',
                       antialiased=False)

ax.set_zticklabels([])

# Generate simplified z-ticks to be displayed on the colorbar instead of the z-axis:
z_min = np.floor(np.min(Z))
z_max = np.ceil(np.max(Z))
z_ticks_simple = np.arange(z_min, z_max+0.01, 0.5)

cbar = ax.figure.colorbar(surf, ax=ax, shrink=0.5, aspect=5)
cbar.ax.set_yticks(z_ticks_simple) # note: here we set the surface's z-ticks as the colorbar's y-ticks
cbar.ax.set_ylabel('Amplitude', rotation=-90, va='bottom')
```
