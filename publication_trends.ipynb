```python
import seaborn as sns
import pandas as pd
```

```python
df = pd.read_csv('dat/pub_trend_example_data_comp_combin_dd.csv')
df.head()
```

```python
df_melt = pd.melt(df, id_vars='year', value_name='n_pubs', var_name='search_kw')
df_melt
```

```python
sns.relplot(
    data=df_melt,
    x='year',
    y='n_pubs',
    hue='search_kw'
)
```

```python
sns.histplot(
    data=df_melt,
    x='year',
    hue='search_kw',
    weights='n_pubs',
    multiple='stack',
    bins=len(df),
    shrink=0.8,
)
```
