# High-Level Statistical Plotting with Seaborn

Seaborn is like the "high-level" counterpart of matplotlib; it allows you to create very complex technical plots in only a few intuitive lines of code. However, seaborn is built on matplotlib, and as such, most plot customization is still performed using matplotlib after initial seaborn plotting.

```python
import seaborn as sns
```

```python
sns.set_theme() # Sets theme to default if none is specified
```

```python
tips = sns.load_dataset('tips') # A standard pandas DataFrame
tips.head()
```

Seaborn's ideology prioritizes the statistical "meaning" or "message" of a plot rather than the explicit plot type itself. As such, the most common type of seaborn plot is the "relationship" plot or `sns.relplot()` that is configured primarily to show the relationship between different aspects of data.

```python
sns.relplot(
    data=tips,
    x='total_bill',
    y='tip'
)
```

```python
sns.relplot(
    data=tips,
    x='total_bill',
    y='tip',
    hue='smoker'
)
```

```python
sns.relplot(
    data=tips,
    x='total_bill',
    y='tip',
    hue='smoker',
    style='smoker'
)
```

```python
sns.relplot(
    data=tips,
    x='total_bill',
    y='tip',
    hue='smoker',
    style='time'
)
```

```python
sns.relplot(
    data=tips,
    x='total_bill',
    y='tip',
    hue='size'
)
```

```python
sns.relplot(
    data=tips,
    x='total_bill',
    y='tip',
    size='size'
)
```

```python
sns.relplot(
    data=tips,
    x='total_bill',
    y='tip',
    col='time',
    hue='smoker',
    size='size'
)
```

```python
sns.lmplot(
    data=tips,
    x='total_bill',
    y='tip',
    col='time',
    hue='smoker'
)
```

```python
sns.displot(
    data=tips,
    x='total_bill',
    col='time',
    kde=True # Here we use kernel density estimation, 
             # but default is a standard histogram
)
```

```python
sns.displot(
    data=tips,
    kind='ecdf',
    x='total_bill',
    col='time',
    hue='smoker',
    rug=True
)
```

```python
sns.catplot(
    data=tips,
    kind='swarm',
    x='day',
    y='total_bill',
    hue='smoker'
)
```

A brief recorded discussion of violin and box plots in seaborn can be found [here](https://northeastern.hosted.panopto.com/Panopto/Pages/Viewer.aspx?id=31fe2a08-f893-4b1f-9e82-b471013e24ab)

```python
sns.catplot(
    data=tips,
    kind='violin',
    x='day',
    y='total_bill',
    hue='smoker',
    split=True
)
```

```python
sns.catplot(
    data=tips,
    kind='bar',
    x='day',
    y='total_bill',
    hue='smoker'
)
```
