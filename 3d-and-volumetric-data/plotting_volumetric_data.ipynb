```python
import os

import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from gridData import Grid
import plotly.graph_objects as go
```

```python
volmap = Grid(os.path.join('..', 'dat', '1o7v-sia_global_9999p_intengmin_map.dx'))

data_3d = volmap.grid
print(f"Volumetric map data dimensions (X, Y, Z): {data_3d.shape}")
```

```python
print(np.arange(data_3d.shape[0]))
```

```python
print(np.min(data_3d)*0.8)
print(np.max(data_3d))
```

[Built-in colorscales with Plotly](https://plotly.com/python/builtin-colorscales/)

```python
x, y, z = np.indices(data_3d.shape)

fig = go.Figure(data=go.Volume(
    x=x.ravel(),
    y=y.ravel(),
    z=z.ravel(),
    value=data_3d.ravel(),
    isomin=np.min(data_3d)*0.95,
    isomax=np.min(data_3d)*0.5,
    opacity=0.1,
    surface_count=20,
    colorscale='RdPu_r'
))

fig.update_layout(
    scene=dict(
        xaxis=dict(range=[0, data_3d.shape[0]]),
        yaxis=dict(range=[0, data_3d.shape[1]]),
        zaxis=dict(range=[0, data_3d.shape[2]])
    )
)
fig.show()
```
