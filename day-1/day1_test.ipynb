```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
main_text = 'NU COMMLAB'
sub_text = 'Python Data Visualization Workshop'

primary_color = 'teal'
secondary_color = 'paleturquoise'
```

```python
# Define functions for the main speech bubble:
x = np.arange(-32, 32, 0.5)
y1 = np.sin(9*x) + np.sin(9.1*x)
y2 = np.cos(9*x) + np.cos(8.9*x)

# Define functions for the speech bubble tail (bonus points if someone can help me get this function to decay faster!):
x_tail = np.arange(10, 20, 0.5)
y_tail_1 = ((5/x_tail) * (np.sin(5*x_tail)) - 0.75) - (0.05*x_tail)
y_tail_2 = ((5/x_tail) * (np.cos(5*x_tail)) - 0.75) - (0.05*x_tail)

# Define stochastic functions for the central random waveform in the middle:
np.random.seed(81)
def add_noise(y, loc=0, scale=0.01):
    noise = np.random.normal(loc=loc, scale=scale, size=y.shape)
    return y + noise
x_noise_left = np.arange(-35, 20, 0.25)
x_noise_right = np.arange(-20, 35, 0.25)
y_noise_left = add_noise(0.25*np.sin(x_noise_left), scale=0.3)
y_noise_right = add_noise(0.25*np.sin(x_noise_right), scale=0.3)

# Setup subplots:
fig, ax = plt.subplots()

# Plot the main speech bubble:
ax.plot(x, y1, color=primary_color)
ax.plot(x, y2, color=secondary_color)

# Plot the speech bubble's tail:
ax.plot(x_tail, y_tail_1, color=primary_color)
ax.plot(x_tail, y_tail_2, color=secondary_color)

# Plot the central waveform:
ax.plot(x_noise_left, y_noise_left, color=primary_color)
ax.plot(x_noise_right, y_noise_right, color=secondary_color)

# Add Graphic text!
ax.text(0, 0.1, s=main_text,
        backgroundcolor='white',
        color=primary_color,
        fontname='Arial',
        fontsize='24',
        fontweight='bold',
        ha='center',
        va='center')
ax.text(0, -0.2, s=sub_text,
        backgroundcolor='white',
        color=primary_color,
        fontname='Arial',
        fontsize='10.2',
        fontweight='bold',
        ha='center',
        va='center')
```
