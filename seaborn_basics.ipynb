# Python Plotting with Seaborn

Seaborn is like the "high-level" counterpart of matplotlib; it allows you to create very complex technical plots in only a few intuitive lines of code. However, seaborn is built on matplotlib, and as such, most plot customization is still performed using matplotlib after initial seaborn plotting.

```python
import seaborn as sns
```

Seaborn supports multiple pre-formatted themes for a variety of plotting types through the [`sns.set_theme()`](https://seaborn.pydata.org/generated/seaborn.set_theme.html) method.

```python
sns.set_theme() # Sets theme to default if none is specified
```

Seaborn's `sns.load_dataset()` method also comes in handy for quickly testing different types of visuals on datasets that come pre-installed with Seaborn. Each of these datasets are in standard Pandas DataFrame format; a complete list of available example datasets can be found [here](https://github.com/mwaskom/seaborn-data).

```python
tips = sns.load_dataset('tips') # A standard pandas DataFrame
tips.head()
```

### Plotting Relationships Between Variables

Seaborn's ideology prioritizes the statistical "meaning" or "message" of a plot rather than the explicit plot type itself. As such, the most common type of seaborn plot is the "relationship" plot or `sns.relplot()` that is configured primarily to show the relationship between different aspects of data.

All we need to output a `relplot` is our input dataset via the `data` keyword argument and to specify which columns will be plotted on either axis via the `x` and `y` keyword arguments. Seaborn will automatically guess what type of plot would be best for our needs; in the case of plotting the `total_bill` vs the `tip` columns, it outputs a standard scatter plot.

```python
sns.relplot(
    data=tips,
    x='total_bill',
    y='tip'
)
```

We can add another dimension by distinguishing smoker vs non-smoker of each sample by color. Seaborn automatically displays this just by specifying a column to the `hue` keyword argument.

```python
sns.relplot(
    data=tips,
    x='total_bill',
    y='tip',
    hue='smoker'
)
```

To provide an additional layer of distinguishment and increase accessiblity, the `style` keyword argument can be used to change the shape of each sample in accordance with same column specified for `hue`. It is possible to specify `style` and `hue` using different features to add yet another dimension, but this can often lead to difficulties in differentiating between color and sample shape.

```python
sns.relplot(
    data=tips,
    x='total_bill',
    y='tip',
    hue='smoker',
    style='smoker'
)
```

While the previous example using `hue` displayed a color binary, continuous values can also be passed to `hue` which will create a color gradient with intensity corresponing to the magnitude of each sample's value in the specified category. Here, we pass the `size` column of our dataset to `hue` to easily show the party size of each sample along with their total bill and tip. As expected, larger parties were more likely to have a larger total bill and to have paid a larger tip.

```python
sns.relplot(
    data=tips,
    x='total_bill',
    y='tip',
    hue='size'
)
```

Seaborn also provides a `size` keyword argument that can scale the diameter of each sample point by a specified dataset feature. By default, the sample point diameter is increased linearly and does not directly correspond to the magnitude of the feature value of each sample.

```python
sns.relplot(
    data=tips,
    x='total_bill',
    y='tip',
    size='size'
)
```

When more than 3 or 4 dimensions need to be compared across a dataset, it is often best to create adjacent plots to compare between. Seaborn provides easy support for this through the `col` keyword argument which sets up a figure with n plots corresponding to each categorical value in the specified feature. Here, we are able to display and compare across 5 different dimensions present in our dataset simultaneously using the x-axis, y-axis, sample point color, sample point size, and two side-by-side plots.

```python
sns.relplot(
    data=tips,
    x='total_bill',
    y='tip',
    col='time',
    hue='smoker',
    size='size'
)
```

The `sns.lmplot()` function takes `relplot()` a step further by bootstrapping a regression model across the data. 

```python
sns.lmplot(
    data=tips,
    x='total_bill',
    y='tip',
    col='time',
    hue='smoker',
)
```

### Visualizing Distributions Across a Dataset

Seaborn's `sns.displot()` can be used to plot three main types of data visuals used to show the distribution of a certain aspect of a dataset: histograms, kernel density estimates (KDEs), and empirical cumulative distribuiton functions (ECDFs). This function can output both univariate and bivariate distribution plots to compare distributions across related features of a dataset.

Just specifying the dataset and data feature using the `data` and `x` keyword arguments gives a simple histogram with automatically calculated bin edges.

```python
sns.displot(
    data=tips,
    x='total_bill',
)
```

Specifying the `kind` keyword argument allows access to KDE and ECDF. A KDE plot can be superimposed on a histogram by instead specifying `kde=True`

```python
sns.displot(
    data=tips,
    x='total_bill',
    kind='kde'
)
```

```python
sns.displot(
    data=tips,
    x='total_bill',
    kde=True 
)
```

```python
sns.displot(
    data=tips,
    x='total_bill',
    kind='ecdf'
)
```

We can also use the same keyword arguments as in `relplot()` and `lmplot()` to add further dimension to the distribution plot. Here we add a `rug` to show sample density along a single axis.

```python
sns.displot(
    data=tips,
    kind='ecdf',
    x='total_bill',
    col='time',
    hue='smoker',
    rug=True
)
```

Bivariate distributions can be displayed using 2d histograms or KDE contours by specifying both the `x` and `y` keyword arguments with separate dataset features. 

```python
sns.displot(
    data=tips,
    x='total_bill',
    y='tip',
)
```

```python
sns.displot(
    data=tips,
    x='total_bill',
    y='tip',
    kind='kde'
)
```

### Visualizing Categorical Data

The `sns.catplot` figure-level function handles several types of categorical data visualizations like bar, box, and violin plots.

```python
sns.catplot(
    data=tips,
    kind='bar',
    x='day',
    y='total_bill',
    hue='smoker'
)
```

Setting the `kind` keyword argument to 'box' or 'violin' gives a visual of the distribution of data in each specified category. A brief recorded discussion of violin and box plots in seaborn can be found [here](https://northeastern.hosted.panopto.com/Panopto/Pages/Viewer.aspx?id=31fe2a08-f893-4b1f-9e82-b471013e24ab)

```python
sns.catplot(
    data=tips,
    kind='box',
    x='day',
    y='total_bill',
    hue='smoker',
)
```

```python
sns.catplot(
    data=tips,
    kind='violin',
    x='day',
    y='total_bill',
    hue='smoker',
    split=True
)
```

Lesser-used visuals like swarmplots can be output via `sns.catplot()` and can be useful for some cases.

```python
sns.catplot(
    data=tips,
    kind='swarm',
    x='day',
    y='total_bill',
    hue='smoker',
)
```

Combining figure-level and axes-level functions in Seaborn allows us to show multiple representations on the same figure. Here, we use the figure-level `catplot()` and axes-level `swarmplot()` functions to overlay a swarmplot onto a background violin plot.

```python
sns.catplot(
    data=tips,
    kind='violin',
    x='day',
    y='total_bill',
    color='.9',
    inner=None
)

sns.swarmplot(
    data=tips,
    x='day',
    y='total_bill',
    hue='smoker'
)
```
